# Dealing with missing values



# 1. Droping null values

dropna()-->>removes row if it has a null value

dropna(how='any')-->>same as dropna()

dropna(how='all')-->>removes a row if it has all the values are null

dropna(axis=1)--->>>removes the column with null values

```python
df.dropna()
```

```python
df.dropna(how='any')
```

```python

df.dropna(how='all')
```

```python

df.dropna(axis=1)
```

#2. Filling missing values

1.using fillna() with mean(),median(),mode()

2.using forward fill and backward fill

```python
df['age']
```

```python
#mean()
df['age'].fillna(df['age'].mean())
```

```python
#median
df['age'].fillna(df['age'].median())
```

```python
#mode
df['age'].fillna(df['age'].mode())
```

```python
# drawback is if the first value null then null is filled in that
df['age'].fillna(method='ffill')
```

```python
#backword fill
# drawback is if the last value null then null is filled in that
df['age'].fillna(method='bfill')
```

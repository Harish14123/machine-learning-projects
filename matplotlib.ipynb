```python
#univarient-->>focusing on only one column(numerical)
data={
    'salary':[10000,15000,20000,25000,30000,25000,20000,50000,50000,55000,60000,65000,70000,75000,30000,35000]
}
```

```python
import pandas as pd
df=pd.DataFrame(data)
df.head()
```

```python
plt.plot(df['salary'])
plt.show()
```

```python
from matplotlib import lines
plt.plot(df['salary'],color='blue',marker='o',linestyle='--',linewidth=2)
plt.grid()
plt.show()
```

# plotting a histogram

```python
plt.hist(df['salary'])
plt.show()
```

# Box plot

```python
plt.boxplot(df['salary'])
plt.show()
```

# Univarient categorical column

```python
df.shape
```

```python
df['dept']=['it','hr','finance','sales','it','hr','finance','sales','it','hr','finance','sales','it','hr','finance','sales']
df.head()
```

# pie chart is used to visualize categorical data

```python
count=df['dept'].value_counts()
count
```

```python
#labels for naming the section
#autopct is use to represent percentage
#explode for seperating a block from the pie
# axis('equal) makes a perfect circle
plt.pie(count,labels=count.index,autopct='%1.2f',explode=[0.1,0,0,0])
plt.axis('equal')
plt.show()
```

# Bar chart Using count

```python
plt.bar(count.index,count)
plt.show()
```

# Bivarient(two variables)
1.numeric to numeric

2.numeric to categorical

```python

df['age']=[22,24,30,32,25,26,41,34,52,43,42,21,22,20,22,33]
df.head()
```

# Scatter plot

```python
plt.scatter(df['salary'],df['age'])
plt.show()
```

# line plot

```python
sorage=df['age'].sort_values()
plt.plot(df['salary'],sorage)
plt.show()
```

# bar chart

```python
plt.bar(df['age'],df['salary'])
plt.show()
```

```python
# numeric-categerical

it_sal=df[df['dept']=='it']['salary']
hr_sal=df[df['dept']=='hr']['salary']
finance_sal=df[df['dept']=='finance']['salary']
sales_sal=df[df['dept']=='sales']['salary']
```

```python
#box plot
plt.boxplot([it_sal,hr_sal,finance_sal,sales_sal],labels=['it','hr','finance','sales'])
plt.show()
```

```python
#pie plot
sal_by_dept=df.groupby('dept')['salary'].mean()
sal_by_dept
```

```python
plt.pie(sal_by_dept,labels=sal_by_dept.index,autopct='%1.2f')
plt.show()
```

# multivarient analysis-->>three or more variables

```python
#using bubble plot
df['expe']=[1,2,3,4,1.5,5,4.5,3,2,1,6,3.5,4,3,2,1]
df.head()
```

```python
plt.scatter(df['salary'],df['age'],s=df['expe']*100)
plt.show()
```

# obeject oriented api

```python
fig,axes=plt.subplots(1,3,figsize=(15,5))
axes[0].scatter(sorage,df['salary'])
axes[1].boxplot([it_sal,hr_sal,finance_sal,sales_sal],tick_labels=['it','hr','finance','sales'])
axes[2].pie(sal_by_dept,labels=sal_by_dept.index,autopct='%1.2f')
plt.tight_layout()
plt.savefig('threedig.png')
plt.show()
```

```python
df2={
    'year':[2020,2021,2022,2023],
    'sales':[100,150,200,250],
    'profit':[30,40,50,60],
    'Expenses':[80,120,160,200]
}
```

```python
plt.plot(df2['year'],df2['sales'])
plt.plot(df2['year'],df2['profit'])
plt.plot(df2['year'],df2['Expenses'])
plt.title('Financial Analysis')
plt.xlabel('Year')
plt.ylabel('Amount')
plt.legend(['sales','profit','Expenses'])
plt.show()
```

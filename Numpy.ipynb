# reshaping-->>reshape(),ravel(),flatten()

reshape is only successful when there are sufficient elements for reshape

ravel gives the view of the original array so changes in the ravel array will change original array

flatten gives the copy of it

```python
a=np.arange(1,10)
a.reshape(3,3)

r=a.ravel()
r[0]=10
print(r)
print(a)
print()
f=a.flatten()
f[0]=100
print(a)
print(f)
```

# conditional based choices

1.where

2.argwhere-->>gives the position where the condition is satisfied

3.nonzero-->>gives the indexs of non zero indices

4.np.logical_and

5.np.logical_or

```python
a=np.array([2,3,7,8,5])
print(np.where(a<5,'f','t'))
```

```python
print(np.argwhere(a<5))
```

```python
b=np.array([0,3,0,4,0,6,7])
print(np.nonzero(b))
```

```python
a=np.array([1,2,3,4,5,6])
mask=np.logical_and(a>2,a<6)
print(mask)
print(a[mask])
```

```python
mask=np.logical_or(a>2,a<6)
print(mask)
print(a[mask])
```

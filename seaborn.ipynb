```python
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
sns.get_dataset_names()
```

```python
df=sns.load_dataset('penguins')
df.head()
```

```python
df['species'].value_counts()
```

```python
df['sex'].value_counts()
```

# Scatter plot using seaborn

1.set_style()-->>white,dark,whitegrid,darkgrid,ticks

2.despine()--->>> removing the left or bottom or any axis line

```python
sns.set_style('ticks')
sns.scatterplot(data=df,x='flipper_length_mm',y='body_mass_g',hue='species')
sns.despine()
```

# context setting
1.default notebook

2.paper

3.talk

4.poster

```python
sns.set_style('darkgrid')
sns.scatterplot(data=df,x='flipper_length_mm',y='body_mass_g',hue='species')
sns.despine()
```

```python
sns.set_style('ticks')
sns.set_context('paper')
sns.scatterplot(data=df,x='flipper_length_mm',y='body_mass_g',hue='species')
sns.despine()
```

```python
sns.set_style('ticks')
sns.set_context('talk')
sns.scatterplot(data=df,x='flipper_length_mm',y='body_mass_g',hue='species')
sns.despine()
```

```python
sns.set_style('ticks')
sns.set_context('poster')
sns.scatterplot(data=df,x='flipper_length_mm',y='body_mass_g',hue='species')
sns.despine()
```

# bar plot

```python
sns.barplot(data=df,x='species',y='body_mass_g',hue='sex',palette=['pink','skyblue'])
```

# heatmap
used for visualizing the correaltions between numerical variables

used for visualizing confusion matrix

used for highlighting missing value

```python
#annot used for displaying the values in the boxes
# cmap is the color of the map
col=['bill_length_mm','bill_depth_mm','flipper_length_mm','body_mass_g']
sns.heatmap(data=df[col].corr(),annot=True,cmap='coolwarm')
```

# pair plot

```python
sns.pairplot(df,hue='island')
```

#pairgrid plot-->>it is a subplotgrid used to represent the pairwise relationships between variables

```python
graph=sns.PairGrid(data=df,hue='sex',palette='Set2')
graph.map_upper(sns.scatterplot)
graph.map_lower(sns.kdeplot)
graph.map_diag(sns.histplot,kde=True)
```
